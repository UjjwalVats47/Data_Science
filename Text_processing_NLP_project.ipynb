```python
import numpy as np
import pandas as pd
```

```python
with open("text preprocessing data 1.txt", 'r') as f:
    lines = f.readlines()
df = pd.DataFrame(lines, columns=['text_content'])
display(df.head())
```

```python
df.shape
```

```python
df['text_content'] = df['text_content'].str.lower()
```

```python
df
```

Removing HTML Tags

```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

```python
df['text_content'] = df['text_content'].apply(remove_html_tags)
```

```python
df['text_content'][15]
```

Removing link text

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

```python
df['text_content']= df['text_content'].apply(remove_url)
```

```python
df['text_content']
```

Removing Bracketed Numbers/Citations

```python
df['text_content'] = df['text_content'].str.replace(r'\[\d+\]', '', regex=True)
```

Removing \n

```python
df['text_content'] = df['text_content'].str.replace('\n',' ')
```

Removing punctuation

```python
import string,time
string.punctuation
```

```python


exclude = string.punctuation
```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,' ')
    return text
```

```python
df['text_content'] = df['text_content'].apply(remove_punc)
```

```python
df['text_content'][1]
```

Removing Stop words

```python
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
nltk.download ("stopwords")
```

```python
stopwords.words('english')
```

```python
st=stopwords.words('english')
```

```python
st
```

```python
df2= ''
```

```python
for text_entry in df['text_content']:
    # Ensuring text_entry is a string before splitting
    if isinstance(text_entry, str):
        words = text_entry.split()
        for word in words:
            if word not in st:
                df2 += word + ' '
```

Removing Inflection(Lemmatization Method)

```python
from nltk.stem import WordNetLemmatizer
wnl= WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
def lemmatize_text(text):
    words = text.split()
    lemmatized_words = [wnl.lemmatize(word) for word in words]
    return ' '.join(lemmatized_words)

lemmatized_df2 = lemmatize_text(df2)
print(lemmatized_df2)
```
